## Visualization 1: Temperature and Dissolved Oxygen

In aquatic systems, warmer water often holds less dissolved oxygen. Let us look for that pattern in this sample dataset.

```python
plt.scatter(df['temperature_c'], df['dissolved_oxygen_mg_l'])
plt.xlabel('Temperature (°C)')
plt.ylabel('Dissolved Oxygen (mg/L)')
plt.title('Temperature vs. Dissolved Oxygen in Sample Louisiana Wetland Sites')
plt.show()
```

## Visualization 2: Chlorophyll-a by Habitat Type

Chlorophyll-a can be used as an indicator of algal biomass. Higher values may suggest greater algal activity, but interpretation depends on context.

```python
df.boxplot(column='chlorophyll_a_ug_l', by='habitat_type', rot=45)
plt.xlabel('Habitat Type')
plt.ylabel('Chlorophyll-a (µg/L)')
plt.title('Chlorophyll-a by Habitat Type')
plt.suptitle('')
plt.tight_layout()
plt.show()
```

### Interpretation Prompt

Which habitat type appears to have higher chlorophyll-a values? What might explain this pattern? What additional data would help you avoid overclaiming?

## Visualization 3: Nutrients and Algal Cell Count

Nutrients such as nitrate can influence algal growth. Let us compare nitrate concentration and algal cell count.

```python
plt.scatter(df['nitrate_mg_l'], df['algal_cell_count_ml'])
plt.xlabel('Nitrate (mg/L)')
plt.ylabel('Algal Cell Count (cells/mL)')
plt.title('Nitrate vs. Algal Cell Count')
plt.show()
```
